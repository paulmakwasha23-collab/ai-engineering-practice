```python
documents = [
    "Paracetamol is commonly used to relieve mild to moderate pain and reduce fever.",
    "Ibuprofen is a non-steroidal anti-inflammatory drug used for pain, inflammation and fever.",
    "Amoxicillin is a beta-lactam antibiotic used to treat certain bacterial infections.",
    "Metformin is commonly used as an oral medicine for type 2 diabetes.",
    "Salbutamol is a short-acting bronchodilator commonly used to relieve bronchospasm."
]

for document in documents:
  print(document)
```

```python
query = "Which medicine treats bacterial infections?"
print(query)
```

```python
for document in documents:
  print(document)
  print()
```

```python
for document in documents:
  if "bacterial" in document:
    print(document)
```

```python
keywords = query.split()
print(keywords)
```

```python
for document in documents:
  for word in keywords:
    if word.lower() in document.lower():
      print(document)
      break
```

```python
for document in documents:
  matches = 0
  for word in keywords:
    if word.lower() in document.lower():
      matches += 1
  print(matches,document)
```

```python
for document in documents:
  matches = []
  for word in keywords:
    if word.lower() in document.lower():
      matches.append(word)
  print(matches,document)
```
